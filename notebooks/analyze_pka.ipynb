# Analyze an Atomify PKA run
This notebook reads `pka_metrics.dat` produced by `PKA_O.in` or `PKA_U.in`. It separates **peak disturbance** from **residual damage**, which is the central idea of the exercise.


```python
import numpy as np
import matplotlib.pyplot as plt

DT_PS = 0.00025  # PKA-stage timestep from the LAMMPS input
data = np.loadtxt('pka_metrics.dat', comments='#')
step = data[:, 0]
time_ps = step * DT_PS
msd = data[:, 1]
n_displaced = data[:, 2]
max_disp = data[:, 3]
pka_ke = data[:, 4]

print(f'{len(time_ps)} recorded frames, ending at {time_ps[-1]:.3f} ps')
```

```python
fig, ax = plt.subplots()
ax.plot(time_ps, msd)
ax.set_xlabel('Time after PKA (ps)')
ax.set_ylabel('Mean-square displacement (Angstrom$^2$)')
ax.set_title('Whole-crystal disturbance')
plt.show()
```

```python
fig, ax = plt.subplots()
ax.plot(time_ps, n_displaced)
ax.set_xlabel('Time after PKA (ps)')
ax.set_ylabel('Atoms displaced > 1 Angstrom')
ax.set_title('Transient versus residual damage')
plt.show()
```

```python
fig, ax = plt.subplots()
ax.plot(time_ps, max_disp)
ax.set_xlabel('Time after PKA (ps)')
ax.set_ylabel('Maximum atomic displacement (Angstrom)')
ax.set_title('Most strongly displaced atom')
plt.show()
```

```python
fig, ax = plt.subplots()
ax.plot(time_ps, pka_ke)
ax.set_xlabel('Time after PKA (ps)')
ax.set_ylabel('PKA kinetic energy (eV)')
ax.set_title('Energy leaves the primary knock-on atom')
plt.show()
```

```python
peak_i = int(np.argmax(n_displaced))
summary = {
    'peak_displaced_atoms': int(round(n_displaced[peak_i])),
    'time_of_peak_ps': float(time_ps[peak_i]),
    'residual_displaced_atoms': int(round(n_displaced[-1])),
    'peak_MSD_A2': float(np.max(msd)),
    'final_MSD_A2': float(msd[-1]),
    'max_atomic_displacement_A': float(np.max(max_disp)),
}
for key, value in summary.items():
    print(f'{key:28s}: {value}')
```
